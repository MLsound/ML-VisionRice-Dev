### Init

```python
import os
import cv2
```

```python
def convert_yolo_to_csv(image_dir, annotations_dir, output_csv):
    """Convierte anotaciones YOLO TXT a formato CSV para keras-retinanet."""

    with open(output_csv, 'w') as csv_file:
        csv_file.write("filename,x1,y1,x2,y2,class_name\n")  # Encabezado CSV

        for filename in os.listdir(image_dir):
            if filename.endswith(('.jpg', '.jpeg', '.png')):
                image_path = os.path.join(image_dir, filename)
                txt_filename = os.path.splitext(filename)[0] + '.txt'
                txt_path = os.path.join(annotations_dir, txt_filename)

                if os.path.exists(txt_path):
                    img = cv2.imread(image_path)
                    height, width, _ = img.shape

                    with open(txt_path, 'r') as txt_file:
                        for line in txt_file:
                            class_id, x_center, y_center, w, h = map(float, line.split())

                            x1 = int((x_center - w / 2) * width)
                            y1 = int((y_center - h / 2) * height)
                            x2 = int((x_center + w / 2) * width)
                            y2 = int((y_center + h / 2) * height)

                            csv_file.write(f"{image_path},{x1},{y1},{x2},{y2},plant-weed\n")
                else:
                    csv_file.write(f"{image_path},,,,\n")
```

### Set folder structure

```python
train_images_dir = 'train/images/'
train_annotations_dir = 'train/labels/'
val_images_dir = 'valid/images/'
val_annotations_dir = 'valid/labels/'
```

# Convert format

```python
# Define los archivos de salida
train_output_csv = 'train_annotations.csv'
val_output_csv = 'validation_annotations.csv'
```

```python
# Convertir anotaciones
convert_yolo_to_csv(train_images_dir, train_annotations_dir, train_output_csv)
convert_yolo_to_csv(val_images_dir, val_annotations_dir, val_output_csv)

print("✅ Conversión completada.")
```
